```python
from ultralytics import YOLO
```

```python
yolo_model = YOLO("yolov8n.pt") 
```

```python
results = yolo_model.predict(
            source="/home/cyberhunter/Ai-Proctor/AI-Model/debug_faces/3_566.jpg", 
            conf=0.5,           # Higher confidence threshold
            iou=0.7,            # Higher IoU threshold to reduce overlapping boxes
            verbose=False,
            device='cpu',       # Explicitly use CPU
            half=False,         # Disable half precision for CPU
            max_det=10          # Limit max detections per image
        )
```

```python
print(results)
```

```python
detected_objects = {"person": 0, "cell phone": 0, "laptop": 0}
```

```python
print(len(results))
```

```python
for box in results[0].boxes:
    label = yolo_model.names[int(box.cls[0])]
    if label in detected_objects:
        detected_objects[label] += 1
```
